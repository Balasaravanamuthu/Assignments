```python
import pandas as pd
import pickle
from sklearn.pipeline import Pipeline
dataset = pd.read_csv("insurance_pre.csv")
```

```python
dataset
```

```python
dataset = pd.get_dummies(dataset, drop_first = True).astype(int)
```

```python
dataset
```

```python
dataset.columns
```

```python
independent = dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
independent
```

```python
dependent = dataset[['charges']]
dependent
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(independent, dependent, test_size=1/3, random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
#sc = StandardScaler()
#X_train = sc.fit_transform(X_train)
#X_test = sc.transform(X_test)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('svr', SVR())
])

param_grid = {'svr__kernel': ['rbf', 'poly', 'linear'], 'svr__C': [0.1, 1, 10, 100, 1000],
    'svr__epsilon': [0.001, 0.01, 0.1, 0.5, 1],
    'svr__gamma': ['scale', 'auto']}


#grid = GridSearchCV(SVR(), param_grid, refit = True, verbose = 3, n_jobs=1)
grid = GridSearchCV(estimator=pipeline, param_grid=param_grid,refit = True, verbose = 3, n_jobs=1)
grid.fit(X_train, y_train)
```

```python
print("The R score value for the best parameter {}".format(grid.best_params_))
```

```python
re = grid.cv_results_
```

```python
table = pd.DataFrame.from_dict(re)
```

```python
import pickle
```

```python
filename = "finalisedmodel.sav"
```

```python
age_input=float(input("Age=:"))
bmi_input=float(input("BMI=:"))
children_input=float(input("Children=:"))
sex_male_input=int(input("Sex=:"))
smoker_yes_input=int(input("Smoker=:"))
preinput = [[age_input,bmi_input,children_input,sex_male_input,smoker_yes_input]]
```

```python
loaded_model = pickle.load(open(filename,"rb"))
result = loaded_model.predict(preinput)
result
```
